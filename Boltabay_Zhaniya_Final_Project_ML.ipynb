```python
#Создание копии данных
data = df.copy()
#Удаление ненужных столбцов
data = data.drop(columns=['Patient_ID', 'Overall_Risk_Score', 'Risk_Level'])
#Закодирование типов рака в бинарные числа
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y = le.fit_transform(data['Cancer_Type'])
#Проверка соотвествий
mapping = dict(zip(le.classes_, le.transform(le.classes_)))
print("Числовой:", mapping)
print("Оригинал:", le.inverse_transform([0, 1, 2]))
```

```python
#Удаление первичного(оригинал) признака
X = data.drop(columns=['Cancer_Type'])
```

```python
#Разделение на тестовый/тренировачный выборки
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y)
```

```python
#Масштабирование числовых признаков
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled  = scaler.transform(X_test)
```

```python
#Логистическая регрессия
from sklearn.linear_model import LogisticRegression
model_lr = LogisticRegression(multi_class='multinomial', max_iter=1000, random_state=42)
model_lr.fit(X_train_scaled, y_train)
y_pred_lr = model_lr.predict(X_test_scaled)
y_prob_lr = model_lr.predict_proba(X_test_scaled)
```

```python
#K-ближайших соседей
from sklearn.neighbors import KNeighborsClassifier
model_knn = KNeighborsClassifier(n_neighbors=5)
model_knn.fit(X_train_scaled, y_train)
y_pred_knn = model_knn.predict(X_test_scaled)
y_prob_knn = model_knn.predict_proba(X_test_scaled)
```

```python
#Случайный лес
from sklearn.ensemble import RandomForestClassifier
model_rf = RandomForestClassifier(n_estimators=100, random_state=42)
model_rf.fit(X_train, y_train)
y_pred_rf = model_rf.predict(X_test)
y_prob_rf = model_rf.predict_proba(X_test)
```

```python
#XGBoost
from xgboost import XGBClassifier
model_xgb = XGBClassifier(objective='multi:softmax', num_class=5,
                          eval_metric='mlogloss', use_label_encoder=False, random_state=42)
model_xgb.fit(X_train, y_train)
y_pred_xgb = model_xgb.predict(X_test)
y_prob_xgb = model_xgb.predict_proba(X_test)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report, confusion_matrix, roc_auc_score

models = {'LogReg': (y_pred_lr, y_prob_lr),
          'KNN':    (y_pred_knn, y_prob_knn),
          'RandomForest': (y_pred_rf, y_prob_rf),
          'XGBoost': (y_pred_xgb, y_prob_xgb)}

metrics = {}
for name, (y_pred, y_prob) in models.items():
    acc = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred, average='weighted')
    rec = recall_score(y_test, y_pred, average='weighted')
    f1 = f1_score(y_test, y_pred, average='weighted')

    try:
        roc = roc_auc_score(pd.get_dummies(y_test), y_prob, multi_class='ovr', average='weighted')
    except Exception:
        roc = None
    metrics[name] = (acc, prec, rec, f1, roc)


for name, (acc, prec, rec, f1, roc) in metrics.items():
    print(f"{name}: Accuracy={acc:.3f}, Precision={prec:.3f}, Recall={rec:.3f}, F1={f1:.3f}, ROC-AUC={roc:.3f}")
```

```python
import pandas as pd
res_df = pd.DataFrame({
    'Model': list(metrics.keys()),
    'Accuracy': [m[0] for m in metrics.values()],
    'F1': [m[3] for m in metrics.values()]
})
display(res_df)
sns.barplot(x='Model', y='Accuracy', data=res_df); plt.title('Accuracy моделей'); plt.show()
sns.barplot(x='Model', y='F1', data=res_df); plt.title('F1-score моделей'); plt.show()
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
plt.figure(figsize=(10,4))
for i,(name,y_pred) in enumerate([('RF', y_pred_rf), ('XGB', y_pred_xgb)]):
    cm = confusion_matrix(y_test, y_pred)
    disp = ConfusionMatrixDisplay(cm, display_labels=le.classes_)
    plt.subplot(1,2,i+1)
    disp.plot(cmap='Blues', values_format='d', ax=plt.gca())
    plt.title(f'Confusion Matrix ({name})')
plt.show()
```

```python
from sklearn.model_selection import GridSearchCV

rf = RandomForestClassifier(random_state=42)
rf_params = {
    'n_estimators': [100, 200],
    'max_depth': [None, 10, 20],
    'min_samples_split': [2, 5],
    'min_samples_leaf': [1, 2]
}
rf_grid = GridSearchCV(estimator=rf, param_grid=rf_params, cv=5, scoring='f1_weighted', n_jobs=-1)
rf_grid.fit(X_train, y_train)
print("Лучшие параметры RF:", rf_grid.best_params_)
best_rf = rf_grid.best_estimator_

y_pred_rf_best = best_rf.predict(X_test)
print("RF (best) Accuracy:", accuracy_score(y_test, y_pred_rf_best))
print("RF (best) F1:", f1_score(y_test, y_pred_rf_best, average='weighted'))
```

```python
rf_compare = pd.DataFrame({
    'Model': ['RF (до)', 'RF (после)'],
    'Accuracy': [accuracy_score(y_test, y_pred_rf), accuracy_score(y_test, y_pred_rf_best)],
    'F1': [f1_score(y_test, y_pred_rf, average='weighted'), f1_score(y_test, y_pred_rf_best, average='weighted')]
})
display(rf_compare)
```

```python
sns.barplot(x='Model', y='F1', data=rf_compare); plt.title('RF до/после tuning'); plt.show()
```

```python
xgb = XGBClassifier(objective='multi:softmax', num_class=5, use_label_encoder=False, eval_metric='mlogloss', random_state=42)
xgb_params = {
    'n_estimators': [100, 200],
    'max_depth': [3, 5, 7],
    'learning_rate': [0.01, 0.1],
    'subsample': [0.8, 1.0]
}
xgb_grid = GridSearchCV(estimator=xgb, param_grid=xgb_params, cv=5, scoring='f1_weighted', n_jobs=-1)
xgb_grid.fit(X_train, y_train)
print("Лучшие параметры XGB:", xgb_grid.best_params_)
best_xgb = xgb_grid.best_estimator_

y_pred_xgb_best = best_xgb.predict(X_test)
print("XGB (best) Accuracy:", accuracy_score(y_test, y_pred_xgb_best))
print("XGB (best) F1:", f1_score(y_test, y_pred_xgb_best, average='weighted'))
```

```python
xgb_compare = pd.DataFrame({
    'Model': ['XGB (до)', 'XGB (после)'],
    'Accuracy': [accuracy_score(y_test, y_pred_xgb), accuracy_score(y_test, y_pred_xgb_best)],
    'F1': [f1_score(y_test, y_pred_xgb, average='weighted'), f1_score(y_test, y_pred_xgb_best, average='weighted')]
})
display(xgb_compare)
```

```python
sns.barplot(x='Model', y='F1', data=xgb_compare); plt.title('XGB до/после tuning'); plt.show()
```
